```python
import pandas as pd

df = pd.read_csv('/content/netflix.csv')
display(df.head())
```

```python
df.info()
```

```python
unique_content_types = df['content_type'].unique()
print("Unique Content Types:")
display(unique_content_types)
```

```python
most_used_device = df['device'].value_counts().idxmax()
print(f"The device used the most times for watching is: {most_used_device}")
```

```python
max_min_watched = df['watched_minutes'].max()
min_min_watched = df['watched_minutes'].min()

print(f"The maximum number of minutes watched in a single session is: {max_min_watched}")
print(f"The minimum number of minutes watched in a single session is: {min_min_watched}")
```

```python
premium_subscribers_df = df[df['subscription'] == 'Premium']
most_watched_genre_premium = premium_subscribers_df['genre'].value_counts().idxmax()

print(f"The genre most watched by Premium subscribers is: {most_watched_genre_premium}")
```

```python
most_popular_hour = df['watch_hour'].value_counts().idxmax()
print(f"The most popular hour for watching Netflix is: {most_popular_hour}:00")
```

```python
df['completion_pct'] = (df['watched_minutes'] / df['total_minutes']) * 100
df.loc[df['total_minutes'] == 0, 'completion_pct'] = 0
display(df.head())
```

```python
average_rating_by_gender = df.groupby('gender')['rating'].mean()
print("Average rating by gender:")
display(average_rating_by_gender)
```

```python
most_popular_show_per_city = df.groupby('city')['show_name'].agg(lambda x: x.mode()[0])
display(most_popular_show_per_city)
```

```python
non_english_sessions = df[df['language'] != 'English']

num_non_english_sessions = len(non_english_sessions)
unique_non_english_languages = non_english_sessions['language'].unique()

print(f"Number of sessions with non-English content: {num_non_english_sessions}")
print(f"Languages involved in non-English content sessions: {', '.join(unique_non_english_languages)}")
```
